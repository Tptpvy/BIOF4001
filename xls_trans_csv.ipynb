### xls转xlsx

```python
import os
import pandas as pd
import numpy as np

file_folder = f"./presumed_infection/2.Data/New 2022" ###Change path
xlsx_folder = os.path.join(file_folder, 'xlsx')

os.makedirs(xlsx_folder, exist_ok=True)

for filename in os.listdir(file_folder):
    if filename.endswith('.xls'):
        file_path = os.path.join(file_folder, filename)

        try:
            tables = pd.read_html(file_path)

            for data_df in tables:
                header_row_idx = data_df.apply(lambda x: x.str.contains('Reference Key', na=False)).any(axis=1).idxmax()
                data_df.columns = data_df.iloc[header_row_idx]
                data_df = data_df[header_row_idx + 1:]

                data_df.dropna(how='all', inplace=True)
                data_df.reset_index(drop=True, inplace=True)

                output_path = os.path.join(xlsx_folder, filename.replace('.xls', '.xlsx'))
                data_df.to_excel(output_path, index=False)

        except Exception as e:
            print(f"处理文件 {filename} 时出错: {e}")


print("所有 xls 文件已转换为 xlsx 格式。")
```

### xlsx转csv

```python
import os
import pandas as pd
file_folder = f"./model/data/lab/" ###Change path
csv_folder = os.path.join(file_folder, 'csv_file')
os.makedirs(csv_folder, exist_ok=True)

for filename in os.listdir(file_folder):
    if filename.endswith('.xlsx'):
        try:
            data_df = pd.read_excel(os.path.join(file_folder, filename),header= None)
            header_row = data_df.apply(lambda x: x.str.contains('Reference Key', na=False).any(), axis=1).idxmax()
            data_df.columns = data_df.iloc[header_row]
            data = data_df.loc[header_row+1:].reset_index(drop=True)
            data = data.dropna(subset=['Reference Key'], how='all')
            csv_path = os.path.join(csv_folder, filename.replace('.xlsx', '.csv'))
            data.to_csv(csv_path, index=False)
        except Exception as e:
            print(f"Error processing file {filename}: {e}")

print("All .xlsx files have been converted to CSV format.")
```

```python
import glob

OUTPUT_PATH = f"./presumed_infection/2.Data/Vincent_topup2/csv_file/" ###Change path
pattern = "*resumedInfection_Bili_Plt_"
year = 2022
files = glob.glob(OUTPUT_PATH + f"{pattern}{year}*.csv")
for file in files:
    df = pd.read_csv(file)
    bili_mask = df['LIS Test Description'].str.contains('bili', case=False, na=False)
    plt_mask = ~bili_mask
    bili_df = df[bili_mask]
    plt_df = df[plt_mask]
    bili_df.to_csv(file.replace('.csv', '_bili.csv'), index=False)
    plt_df.to_csv(file.replace('.csv', '_plt.csv'), index=False)
```
